```python
import torch
import librosa
import gradio as gr

from transformers import pipeline

print("Libraries imported successfully!")
```

```python
device = 0 if torch.cuda.is_available() else -1

emotion_classifier = pipeline(
    "audio-classification",
    model="superb/hubert-base-superb-er",
    device=device
)

print("Speech emotion model loaded successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

audio_file = next(iter(uploaded))

print("Uploaded:", audio_file)
```

```python
results = emotion_classifier(audio_file)

for result in results:
    print(
        result["label"],
        "->",
        round(result["score"] * 100, 2),
        "%"
    )
```

```python
def detect_emotion(audio_file):

    if audio_file is None:
        return "Please upload an audio file."

    results = emotion_classifier(audio_file)

    top_result = results[0]

    emotion_map = {
        "hap": "Happy",
        "sad": "Sad",
        "ang": "Angry",
        "neu": "Neutral"
    }

    label = top_result["label"]
    emotion = emotion_map.get(label, label)

    confidence = top_result["score"]

    return (
        f"Emotion: {emotion}\n"
        f"Confidence: {confidence:.2%}"
    )
```

```python
print(detect_emotion(audio_file))
```

```python
with gr.Blocks() as app:

    gr.Markdown("# 🎭 Speech Emotion Detection")

    gr.Markdown(
        "Upload a speech recording and the AI will "
        "predict the emotion expressed in the audio."
    )

    audio_input = gr.Audio(
        sources=["upload"],
        type="filepath",
        label="Upload Speech"
    )

    detect_button = gr.Button("🎭 Detect Emotion")

    output = gr.Textbox(
        label="Prediction"
    )

    detect_button.click(
        fn=detect_emotion,
        inputs=audio_input,
        outputs=output
    )

app.launch()
```
